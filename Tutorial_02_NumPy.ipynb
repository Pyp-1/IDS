```python
import numpy as np

# Creating a specific 2x2 array with integer elements
array_2x2 = np.array([[10, 20],
                      [30, 40]])

print("--- Task 1: 2x2 Array ---")
print(array_2x2)
print("Shape:", array_2x2.shape)
print("Data Type:", array_2x2.dtype)
```

```python
# Creating a 3x4 array with random floats drawn from a uniform distribution [0, 1)
random_3x4 = np.random.rand(3, 4)

print("\n--- Task 2: Random 3x4 Array ---")
print(random_3x4)
print("Dimensions:", random_3x4.ndim)
print("Shape:", random_3x4.shape)
```

```python
# Creating a 3x3 array filled entirely with ones
array_ones = np.ones((3, 3), dtype=float)

# Creating a 3x3 array filled entirely with zeros
array_zeros = np.zeros((3, 3), dtype=int)

print("\n--- Task 3: Arrays of Ones and Zeros ---")
print("Array of Ones:\n", array_ones)
print("\nArray of Zeros:\n", array_zeros)
```

```python
# Creating a 6x3 random array
original_6x3 = np.random.randn(6, 3)

# Transposing the array
transposed_3x6 = original_6x3.T

print("\n--- Task 4: Transposition ---")
print("Original Shape:", original_6x3.shape)
print("Transposed Shape:", transposed_3x6.shape)
print("\nTransposed Matrix Preview:\n", transposed_3x6)
```

```python
# Define dimensions
n, p, m = 4, 3, 5

# Create two random matrices with matching inner dimensions
matrix_A = np.random.rand(n, p)  # Shape: 4 x 3
matrix_B = np.random.rand(p, m)  # Shape: 3 x 5

# Compute dot product
dot_product_result = np.dot(matrix_A, matrix_B)
# Alternatively: dot_product_result = matrix_A @ matrix_B

print("\n--- Task 5: Matrix Dot Product ---")
print("Shape of Matrix A (n x p):", matrix_A.shape)
print("Shape of Matrix B (p x m):", matrix_B.shape)
print("Shape of Resulting Dot Product (n x m):", dot_product_result.shape)
print("\nResulting Matrix:\n", dot_product_result)
```
